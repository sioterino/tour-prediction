## Quick results check

Full analysis is in `03_results_analysis.ipynb`. This is just a sanity check to confirm the run completed and spot obvious outliers.

```python
metrics = pd.read_csv("E:\\code\\tour-prediction\\data\\results\\metrics.csv")

# Format for readability
display(
    metrics
    .sort_values("R2", ascending=False)
    .reset_index(drop=True)
    .style
    .format({"MAE": "{:,.0f}", "RMSE": "{:,.0f}", "R2": "{:.4f}"})
    .background_gradient(subset=["R2"], cmap="RdYlGn")
    .background_gradient(subset=["MAE", "RMSE"], cmap="RdYlGn_r")
)

print(f"\nTotal combinations run: {len(metrics)} / 18")
print(f"Best R²: {metrics['R2'].max():.4f}  ({metrics.loc[metrics['R2'].idxmax(), 'model']} on {metrics.loc[metrics['R2'].idxmax(), 'dataset']})")
print(f"Worst R²: {metrics['R2'].min():.4f}  ({metrics.loc[metrics['R2'].idxmin(), 'model']} on {metrics.loc[metrics['R2'].idxmin(), 'dataset']})")
```
